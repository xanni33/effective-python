Dopisz definicje dekoratora _timeit_, który wypisze na wyjscie standardowe czas wywołania udekorowanej funckji np.

```
@timeit
def foo(x):
    return x**2
    
r = foo(2)
```
wypisze: <br>
    `Function foo took: 0.00001 seconds` <br>
oraz zwróci 4

````python
from time import time

def timeit(func):
    """ Wypisuje czas wywołania udekorowanej funckji """
    def wrapper(*args, **kwargs):
        start = time()
        result = func(*args, **kwargs)
        end = time()

        print(f"Function {func.__name__} took: {end - start:.5f} seconds")
        return result
    return wrapper

@timeit
def squares_list(n):
    squares = []
    for i in range(n):
        squares.append(i ** 2)
    return squares

@timeit
def squares_comprehension(n):
    return [i ** 2 for i in range(n)]

@timeit
def squares_map(n):
    return map(lambda x: x**2, range(n))

n = 1000000
l = squares_list(n)
c = squares_comprehension(n)
m = squares_map(n)
````

## Zadanie 3 (Igor Sieradzki)

Dopisz definicje dekoratora _derivate_ wg. instrukcji w _docstringu_

````python
import sys
sys.float_info.epsilon # epsilon maszynowy

def derivate(epsilon=None):
    """
    Zwraca pochodną funkcji w punkcie,
    wg wzoru:
    f'(x) = [f(x+h) - f(x)] / h

    Jeśli epsilon nie zostanie podany,
    przyjmujemy 1000 * epsilon maszynowy.
    """

    if epsilon is None:
        epsilon = 1000 * sys.float_info.epsilon

    def decorator(func):

        def wrapper(x):
            h = epsilon
            return (func(x + h) - func(x)) / h

        return wrapper

    return decorator
@derivate(0.01)
def f(x):
    return x*x

@derivate(0.00001)
def g(x):
    return x*x*x+3

def test(a, b, eps=1):
    return abs(round(a)-round(b)) < eps

print(test(f(100), 200.0))
print(round(f(0)) == 0.0)

print(test(g(100), 30000.0))
print(round(g(0)) == 0.0)

from random import random
for x in [random()*1000. for _ in range(20)]:
    print(f(x), 2*x, '\t', test(f(x), 2*x))
    print(g(x), 3*x**2, '\t', test(g(x), 3*x**2))
````

## Zadanie 4 (Igor Sieradzki)

Dopisz definicje dekoratora _accepts_ wg. instrukcji w _docstringu_

````python
def accepts(*types):
    """Sprawdza czy udekorowanej funckji zostały podane odpowiednie parametry zdefiniowane 
       w argumentach dekoratora"""
    def decorator(func):
        def wrapper(*args, **kwargs):
            all_args = list(args) + list(kwargs.values())

            for arg, expected_type in zip(all_args, types):
                if not isinstance(arg, expected_type):
                    raise TypeError(
                        "Expected {}, got {}".format(
                            expected_type.__name__,
                            type(arg).__name__
                        )
                    )
            return func(*args, **kwargs)
        return wrapper
    return decorator

@accepts(str)
def capitalize(word):
    return word[0].upper() + word[1:]

print(capitalize('ola') == 'Ola')

try:
    capitalize(2)
except TypeError:
    print(True)

@accepts(float, int)
def static_pow(base, exp):
    return base ** exp 

print(static_pow(2., 2) == 4.)
print(static_pow(2., exp=2) == 4.)
print(static_pow(base=2., exp=2) == 4.)

try:
    static_pow('x', 10)
except TypeError:
    print(True)
    
try:
    static_pow(2, 2.2)
except TypeError:
    print(True)
````

## Zadanie 5 (Igor Sieradzki)

Dopisz definicje dekoratora _returns_ wg. instrukcji w _docstringu_

````python
from functools import wraps

def returns(*types):
    """
    Sprawdza, czy udekorowana funkcja zwraca wartości
    odpowiednich typów zdefiniowanych w parametrach dekoratora.
    """

    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            result = func(*args, **kwargs)

            if len(types) == 1:
                if not isinstance(result, types[0]):
                    raise TypeError(
                        "Expected {}, got {}".format(
                            types[0].__name__,
                            type(result).__name__
                        )
                    )
            else:
                if not isinstance(result, tuple) or len(result) != len(types):
                    raise TypeError("Expected a tuple with correct number of elements")

                for value, expected_type in zip(result, types):
                    if not isinstance(value, expected_type):
                        raise TypeError(
                            "Expected {}, got {}".format(
                                expected_type.__name__,
                                type(value).__name__
                            )
                        )

            return result

        return wrapper

    return decorator

@returns(str)
def str_only_identity(word):
    return word

print(str_only_identity('hello') == 'hello')

try:
    str_only_identity(10)
except TypeError:
    print(True)
    
@returns(int, int)
def split_indices(x):
    return x[0], x[1]

print(split_indices(x=[6,9]) == (6,9))

try:
    split('AB')
except TypeError:
    print(True)
````

## Zadanie 6 (Igor Sieradzki)
Stwórz dekorator cached służący do cachowania wywołań dowolnej funkcji, tzn. chcemy by:
* wywołanie funkcji z określonymi argumentami miało miejsce tylko raz
* funkcja mogła przyjmować dowolną liczbę nazwanych i nienazwanych argumentów
* nie musi reagować poprawnie na domyślne argumenty, tzn. wywołanie funkcji z domyślnymi argumentami a podanie dokładnie takich samych może być traktowane jako dwa różne wywołania
* na opakowanej funkcji można wywołać `.cache_reset()`, który usunie cache z pamięci
* wywołanie `.cache_status()` zwraca string z opisem w postaci: <br>
    `Function FUNCTION_NAME called X times, evaluated Y times`

````python
from random import random
from functools import wraps

def cached(func):
    """Dekorator zapamiętujący wyniki wywołań funkcji."""

    cache = {}
    calls = 0
    evaluated = 0

    @wraps(func)
    def wrapper(*args, **kwargs):
        nonlocal calls, evaluated

        calls += 1

        key = (args, tuple(sorted(kwargs.items())))

        if key in cache:
            return cache[key]

        result = func(*args, **kwargs)

        cache[key] = result
        evaluated += 1

        return result

    def cache_reset():
        nonlocal calls, evaluated

        cache.clear()
        calls = 0
        evaluated = 0

    def cache_status():
        return "Function {} called {} times, evaluated {} times".format(
            func.__name__, calls, evaluated
        )

    wrapper.cache_reset = cache_reset
    wrapper.cache_status = cache_status

    return wrapper

@cached
def foo(x, y=1, z=4):
    return random()
        
print(foo(3) == foo(3))
print(foo(4) == foo(4))
print(foo(3, z=-1, y=3) == foo(3, y=3, z=-1))
print(foo(3) != foo(x=3))
a = foo(3)
foo.cache_reset()
print(a != foo(3))
print(foo.cache_status() == 'Function foo called 10 times, evaluated 5 times')
````

## Zadanie 7 (Krzysztof Hajto)

Napisz dekorator który będzie robić n-krotne złożenie funkcji, gdzie n jest parametrem dekoratora

````python
from functools import wraps

def zlozenie(n):
    def decorator(func):
        @wraps(func)
        def wrapper(x):
            result = x
            for _ in range(n):
                result = func(result)
            return result
        return wrapper
    return decorator

@zlozenie(3)
def f1(x):
    return x+1

@zlozenie(2)
def f2(x):
    return x*x

@zlozenie(5)
def f3(word):
    return "".join(chr(ord(l)+1) for l in word)

print(f1(2)==5)
print(f2(3)==81)
print(f3("alamakota")=="fqfrfptyf")
````

Python nie ma wbudowanej instrukcji switch. Ale posiada anonimowe funkcje oraz słowniki. Zaimplementuj poniższy switch w postaci słownika funkcji.
`
int my_function(x, y) {


    switch(x) {
        case 1: return y*y;
        case 2: return x+y;
        case 3: return x*y;
        case 4: return 0;
    }
}
`

PS. Nigdy nie róbcie tego w faktycznym kodzie :)

````python
def my_function(x, y):
    switch = {
        1: lambda x, y: y * y,
        2: lambda x, y: x + y,
        3: lambda x, y: x * y,
        4: lambda x, y: 0
    }
    return switch[x](x, y)


print(my_function(1,3)==9)
print(my_function(2,4)==6)
print(my_function(3,1)==3)
print(my_function(4,9)==0)
````
